```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset('titanic')
print("Dataset loaded successfully!\n")

print("1. Shape (rows, columns):", df.shape, "\n")

print("2. Columns:", list(df.columns), "\n")

print("3. First 10 rows:")
print(df.head(10), "\n")

print("4. Last 5 rows:")
print(df.tail(), "\n")

print("5. Info:")
df.info()
print()

print("6. Summary statistics:")
print(df.describe(), "\n")

print("7. Missing values per column:")
print(df.isnull().sum(), "\n")

print("8. Mean age:", round(df['age'].mean(), 2), "\n")

print("9. Top 5 oldest passengers:")
print(df.sort_values('age', ascending=False).head(), "\n")

print("10. Passenger count per class:")
print(df['class'].value_counts(), "\n")

top = df.head()
bottom = df.tail()
combined = pd.concat([top, bottom])
print("CONCAT result (first 5 + last 5 rows):")
print(combined, "\n")

left = pd.DataFrame({'id': [1, 2, 3], 'name': ['A', 'B', 'C']})
right = pd.DataFrame({'id': [1, 2, 3], 'score': [90, 85, 70]})
merged = pd.merge(left, right, on='id')
print("MERGE result:")
print(merged)
```
